```python
class GDRegressor:
    def __init__(self, learning_rate , epochs):
        self.lr = learning_rate
        self.epochs = epochs
        self.coef_ = None
        self.intecept_ =None

    def fit(self, X_train ,Y_train):
        self.intecept_ =0
        self.coef_= np.ones(X_train.shape[1])

        for i in range(self.epochs):
            y_hat = self.intecept_ + np.dot(X_train , self.coef_)
            intercept_dr = -2 * np.mean(Y_train - y_hat)
            self.intecept_= self.intecept_ - (self.lr * intercept_dr)

            coef_dr =  np.dot((Y_train -y_hat) , X_train) * (-2/X_train.shape[0])
            self.coef_ = self.coef_ - (self.lr * coef_dr)

    
    def predict(self, X_test):
        return np.dot(X_test, self.coef_) + self.intecept_




       

         
```

```python
gdr = GDRegressor(0.0001 ,1000000)
gdr.fit(X_train , Y_train)
```

```python
from sklearn.metrics import r2_score

y_predict = gdr.predict(X_test)
r2_score(Y_test , y_predict)
```
